# Damping Sweep

Solve $\ddot{x} + 2\gamma\dot{x} + \omega_0^2 x = 0$ with $\omega_0 = 2$ for three damping rates.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 500)
rates = [0.15, 0.40, 1.00]
print(f"{len(t)} samples, damping rates {rates}")
```

```python
for g in rates:
    x = np.exp(-g * t) * np.cos(np.sqrt(4 - g**2) * t)
    plt.plot(t, x, label=f"γ = {g:.2f}")
plt.xlabel("t"); plt.ylabel("x(t)"); plt.legend()
```

Higher damping rates settle faster but **oscillate less**.
